```python
def calcula_percento(valor, percentual=100):
    quanto = valor * percentual / 100
    return quanto

print ('20% de 300 são: {:10.2f}'.format (calcula_percento (300,20)))

print ('100% de 50 são: {:10.2f}'.format (calcula_percento (50)))
```

```python
import math

n = float(input("Digite o número (n): "))

if n < 0:
    print("Você digitou um número negativo.")
elif n == 0:
    print("Raiz pelo método de Newton: 0.0")
    print(f"Raiz pela função math.sqrt: {math.sqrt(0)}")
else:
    b = n

    while True:
        p = (b + (n / b)) / 2.0
        quadrado_p = p ** 2
        if abs(n - quadrado_p) < 0.0001:
            break

        b = p

    raiz_math = math.sqrt(n)

    print(f"Aproximação por Newton: {p:.6f}")
    print(f"Resultado de math.sqrt:  {raiz_math:.6f}")
```
